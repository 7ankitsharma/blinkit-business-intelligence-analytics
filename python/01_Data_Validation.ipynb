```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 20)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
customers = pd.read_csv("../data/raw/customers.csv")
products = pd.read_csv("../data/raw/products.csv")
stores = pd.read_csv("../data/raw/stores.csv")
delivery_partners = pd.read_csv("../data/raw/delivery_partners.csv")
orders = pd.read_csv("../data/raw/orders.csv")
order_items = pd.read_csv("../data/raw/order_items.csv")
payments = pd.read_csv("../data/raw/payments.csv")
coupons = pd.read_csv("../data/raw/coupons.csv")
calendar = pd.read_csv("../data/raw/calendar.csv")
```

```python
datasets = {
    "Customers": customers,
    "Products": products,
    "Stores": stores,
    "Delivery Partners": delivery_partners,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments,
    "Coupons": coupons,
    "Calendar": calendar
}
```

```python
def dataset_overview(df, name):

    print("="*80)
    print(name.upper())
    print("="*80)

    print(f"Rows : {df.shape[0]:,}")
    print(f"Columns : {df.shape[1]}")

    print("\nData Types")
    print(df.dtypes)

    print("\nMissing Values")
    print(df.isnull().sum())

    print("\nDuplicate Rows")
    print(df.duplicated().sum())

    print("\nMemory Usage")
    print(f"{df.memory_usage(deep=True).sum()/1024**2:.2f} MB")

    print("\nFirst Five Rows")
    display(df.head())
```

```python
for name, df in datasets.items():
    dataset_overview(df, name)
```

```python
quality_report = []

for name, df in datasets.items():

    quality_report.append({

        "Table": name,

        "Rows": df.shape[0],

        "Columns": df.shape[1],

        "Missing Values": df.isnull().sum().sum(),

        "Duplicate Rows": df.duplicated().sum(),

        "Memory (MB)": round(df.memory_usage(deep=True).sum()/1024**2,2)

    })

quality_report = pd.DataFrame(quality_report)

quality_report
```

```python
orders["order_datetime"] = pd.to_datetime(orders["order_datetime"])
orders["delivery_datetime"] = pd.to_datetime(orders["delivery_datetime"])

customers["signup_date"] = pd.to_datetime(customers["signup_date"])

calendar["date"] = pd.to_datetime(calendar["date"])
```

```python
orders["order_hour"] = orders["order_datetime"].dt.hour
```

```python
orders["day_name"] = orders["order_datetime"].dt.day_name()
```

```python
orders["month"] = orders["order_datetime"].dt.month_name()
```

```python
orders["is_weekend"] = orders["day_name"].isin(
    ["Saturday", "Sunday"]
)   
```

```python
orders["delivery_status_category"] = np.where(
    orders["order_status"] == "Cancelled",
    "Cancelled",
    np.where(
        orders["actual_delivery_minutes"]
        <=
        orders["promised_delivery_minutes"],
        "On-Time",
        "Delayed"
    )
)
```

```python
order_items = order_items.merge(
    products[["product_id","cost_price"]],
    on="product_id",
    how="left"
)

order_items["profit"] = (
    order_items["unit_price"]
    - order_items["cost_price"]
) * order_items["quantity"]
```

```python
order_items["profit_margin"] = (
    order_items["profit"]
    /
    order_items["total_price"]
) * 100
```

```python
datasets["Order Items"] = order_items
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```
